
-   Code the solution to your assigned problem in Python (code chunk). Note: each problem can be solved more simply if you use an abstract data type that is suitable for that problem. Using that try to find the best time and space complexity solution!


````python
def is_valid_brackets(s: str) -> bool:
    stack = []
    matching = {
        ')': '(',
        '}': '{',
        ']': '['
    }

    for ch in s:
        if ch in matching.values(): # opening bracket
            stack.append(ch)
        elif ch in matching: # closing bracket
            if not stack or stack[-1] != matching[ch]:
                return False
            stack.pop()
        else:
            return False # invalid character

    return len(stack) == 0

# Testing the code based on the provided examples in this Jupyter Notebook and my provided examples
print(is_valid_brackets("([]{})")) # True
print(is_valid_brackets("([)]")) # False
print(is_valid_brackets("()[]{}")) # True
print(is_valid_brackets("[{]}")) # False

print(is_valid_brackets("{[()()]()[]}")) # True
print(is_valid_brackets("{[()]")) # False
````


-   Explain why your solution works


````python
# The solution works because it uses a stack, which naturally enforces the correct order of brackets.
# Opening brackets are pushed onto the stack as they appear. When a closing bracket is encountered,
# the algorithm checks whether it matches the most recent opening bracket on the stack. If it does not,
# the string is invalid.
# By removing matched brackets from the stack and ensuring that the stack is empty at the end,
# the algorithm guarantees that all brackets are correctly matched and properly nested.
# This directly satisfies both validity conditions of the problem.
# To test the solution, I evaluate the code using several examples, as shown below:
print(is_valid_brackets("([]{})")) # True
print(is_valid_brackets("([)]")) # False
print(is_valid_brackets("()[]{}")) # True
print(is_valid_brackets("[{]}")) # False

print(is_valid_brackets("{[()()]()[]}")) # True
print(is_valid_brackets("{[()]")) # False
````
