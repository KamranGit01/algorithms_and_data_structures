  Given the `root` of a binary tree, return all root to leaf paths in any order.

#### Starter Code for Question 2

```python
from typing import List, Optional

# Definition for a binary tree node.
class TreeNode(object):
    def __init__(self, val = 0, left = None, right = None):
        self.val = val
        self.left = left
        self.right = right

def bt_path(root: Optional[TreeNode]) -> List[List[int]]:
    # TODO
    # 1. Check if the tree is empty and return an empty list if it is.
    # 2. Create a list to store all root-to-leaf paths.
    # 3. Traverse the tree starting from the root.
    # 4. Keep adding the current node’s value to the current path.
    # 5. When a leaf node is reached, save the current path.
    # 6. Go back (backtrack) to explore other possible paths.
    
    if root is None:
        return []

    res: List[List[int]] = []
    path: List[int] = []

    def dfs(node: TreeNode) -> None:
        path.append(node.val)

        if node.left is None and node.right is None:
            res.append(path.copy())
        else:
            if node.left is not None:
                dfs(node.left)
            if node.right is not None:
                dfs(node.right)

        path.pop()

    dfs(root)
    return res


# Example 1
root1 = TreeNode(1)
root1.left = TreeNode(2)
root1.right = TreeNode(2)
root1.left.left = TreeNode(3)
root1.left.right = TreeNode(5)
root1.right.left = TreeNode(6)
root1.right.right = TreeNode(7)
print("Example 1 output:")
print(bt_path(root1))

# Example 2
root2 = TreeNode(10)
root2.left = TreeNode(7)
root2.right = TreeNode(9)
root2.right.right = TreeNode(8)
print("\nExample 2 output:")
print(bt_path(root2))
```

```python
# My New Example:
# Input: nums = [0, 7, 0, 0, 2, 9]
# Output: [7, 2, 9, 0, 0, 0]

# Explanation For My New Example:
# The non-zero numbers (7, 2, and 9) stay in the same order, and all the zeros are moved to the end of the list.

# Example 1 That Has Been Provided by The Partner:
# Input: nums = [2, 0, 7, 0, 3]
# Output: [2, 7, 3, 0, 0]

# Explanation For Example 1 of The Partner:
# Trace walkthrough:
# Do the following for the input nums = [2, 0, 7, 0, 3]:
# 1) Read the list from left to right.
# 2) Keep non-zero values in order: 2, 7, 3
# 3) Count zeros: there are 2 zeros.
# 4) Append the zeros to the end.
# The output will be [2, 7, 3, 0, 0].
# The key point is that the non-zero elements keep their original order, and all zeros are moved to the end.
```


-   Copy the solution your partner wrote. 


```python
from typing import List

def move_zeros_to_end(nums: List[int]) -> List[int]:
    write = 0  # index where the next non-zero should go
    
    # First pass: move non-zeros forward in their original order
    for read in range(len(nums)):
        if nums[read] != 0:
            nums[write] = nums[read]
            write += 1
    
    # Second pass: fill the rest with zeros
    for i in range(write, len(nums)):
        nums[i] = 0
    
    return nums
```


-   Explain why their solution works in your own words.


```python
# It makes one pass through the list to copy all non-zero elements forward in their original order.
# It then fills the remaining positions with zeros.
# This guarantees that the relative order of non-zero elements is preserved, and all zeros end up at the end of the list.
```


-   Explain the problem’s time and space complexity in your own words.


```python
# Time complexity is O(n), because the list is scanned twice, each in linear time.
# Space complexity is O(1), because the list is modified in place using only constant extra variables.
```
